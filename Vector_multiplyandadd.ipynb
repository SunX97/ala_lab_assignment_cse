```python
%%writefile vector_module.py
import sys
import random
import math
from typing import Self

class Vec:
    def __init__(self, src=None) -> Self:
        src_list = list(src) if src is not None else []
        if any(not isinstance(x, (int, float)) for x in src_list):
            raise TypeError("All vector elements must be numbers")
        self.elements = src_list

    def _check_vec(self, t: Self):
        """User-defined helper to validate type and dimensions for vector-vector operations."""
        if not isinstance(t, Vec):
            raise TypeError(f"Expected Vec, got {type(t)}") #Type Safety Check
        if len(self.elements) != len(t.elements):
            raise TypeError("Vectors must have identical dimensions") #Dimensionality Check

    def __add__(self, t: Self) -> Self:
        self._check_vec(t)
        return Vec([round(x + y, 5) for x, y in zip(self.elements, t.elements)])# zip-> pair elements of both vectors by matching index

    def __sub__(self, t: Self) -> Self:
        self._check_vec(t)
        return Vec([round(x - y, 5) for x, y in zip(self.elements, t.elements)])

    def __iadd__(self, t: Self) -> Self: # In-Place addition x = x + y round off to 5 decimal places
        self._check_vec(t)
        for i in range(len(self.elements)):
            self.elements[i] = round(self.elements[i] + t.elements[i], 5)
        return self

    def __radd__(self, other) -> Self: # Right hand addition

        self._check_vec(other)
        return other.__add__(self)

    def __neg__(self) -> Self:
        return Vec([round(-x, 5) for x in self.elements])

    def __rmul__(self, scalar: int | float) -> Self:
        if not isinstance(scalar, (int, float)):
            raise TypeError("Scalar must be an int or float")
        return Vec([round(x * scalar, 5) for x in self.elements])

    def __imul__(self, scalar: int | float) -> Self:
        if not isinstance(scalar, (int, float)):
            raise TypeError("Scalar must be an int or float")
        for i in range(len(self.elements)):
            self.elements[i] = round(self.elements[i] * scalar, 5)
        return self

    def __repr__(self) -> str: return repr(self.elements)
    def __len__(self) -> int: return len(self.elements)

    @staticmethod
    def zeros(n: int) -> Self:
        if n <= 0: raise ValueError("Dimension must be > 0")
        return Vec([0.0] * n)

    @staticmethod
    def ones(n: int) -> Self:
        if n <= 0: raise ValueError("Dimension must be > 0")
        return Vec([1.0] * n)

    @staticmethod
    def uniform(n: int) -> Self:
        if n <= 0: raise ValueError("Dimension must be > 0")
        return Vec([random.random() for _ in range(n)])

    def norm(self) -> float:
        return round(math.sqrt(sum(x ** 2 for x in self.elements)), 5)
```

```python
%%writefile test_vector.py
import time
from vector_module import Vec

def run_tests():
    print("\n=================== DEMONSTRATING VECTOR CALCULATIONS ===================")

    # 1. Base Vectors Creation
    v1 = Vec([1.0, 2.0, 3.0])
    v2 = Vec([4.0, 5.0, 6.0])
    print(f"Vector v1 initialized as: {v1}")
    print(f"Vector v2 initialized as: {v2}")

    # 2. Vector Addition
    start = time.perf_counter()
    v_add = v1 + v2
    t_add = time.perf_counter() - start
    print(f"\n[ADDITION] v1 + v2:")
    print(f"  Calculation: {v1} + {v2}")
    print(f"  Result     : {v_add} (Took {t_add:.8f} seconds)")
    assert str(v_add) == "[5.0, 7.0, 9.0]"

    # 3. Vector Subtraction
    start = time.perf_counter()
    v_sub = v2 - v1
    t_sub = time.perf_counter() - start
    print(f"\n[SUBTRACTION] v2 - v1:")
    print(f"  Calculation: {v2} - {v1}")
    print(f"  Result     : {v_sub} (Took {t_sub:.8f} seconds)")
    assert str(v_sub) == "[3.0, 3.0, 3.0]"

    # 4. Vector Negation
    start = time.perf_counter()
    v_neg = -v1
    t_neg = time.perf_counter() - start
    print(f"\n[NEGATION] -v1:")
    print(f"  Calculation: -({v1})")
    print(f"  Result     : {v_neg} (Took {t_neg:.8f} seconds)")
    assert str(v_neg) == "[-1.0, -2.0, -3.0]"

    # 5. Scalar Multiplication (Left-hand)
    start = time.perf_counter()
    v_mul = 2.0 * v1
    t_mul = time.perf_counter() - start
    print(f"\n[SCALAR MULTIPLICATION] 2.0 * v1:")
    print(f"  Calculation: 2.0 * {v1}")
    print(f"  Result     : {v_mul} (Took {t_mul:.8f} seconds)")
    assert str(v_mul) == "[2.0, 4.0, 6.0]"

    # 6. Right-hand Addition (v2 + v1 fallback scenario)
    start = time.perf_counter()
    v_radd = v2.__radd__(v1)
    t_radd = time.perf_counter() - start
    print(f"\n[RIGHT-HAND ADDITION] v2 + v1:")
    print(f"  Calculation: Right addition fallback step")
    print(f"  Result     : {v_radd} (Took {t_radd:.8f} seconds)")

    # 7. In-Place Addition (+=)
    v_iadd_base = Vec([1.0, 2.0, 3.0])
    print(f"\n[IN-PLACE ADDITION] v1 += v2:")
    print(f"  Original v1: {v_iadd_base}")
    start = time.perf_counter()
    v_iadd_base += v2
    t_iadd = time.perf_counter() - start
    print(f"  Modified v1: {v_iadd_base} (Took {t_iadd:.8f} seconds)")

    # 8. In-Place Multiplication (*=)
    v_imul_base = Vec([1.0, 2.0, 3.0])
    print(f"\n[IN-PLACE MULTIPLICATION] v1 *= 3:")
    print(f"  Original v1: {v_imul_base}")
    start = time.perf_counter()
    v_imul_base *= 3
    t_imul = time.perf_counter() - start
    print(f"  Modified v1: {v_imul_base} (Took {t_imul:.8f} seconds)")

    # 9. Euclidean Norm Calculation
    v_norm = Vec([3.0, 4.0])
    start = time.perf_counter()
    n_res = v_norm.norm()
    t_norm = time.perf_counter() - start
    print(f"\n[EUCLIDEAN NORM] v_norm.norm():")
    print(f"  Calculation: sqrt(3.0^2 + 4.0^2)")
    print(f"  Result     : {n_res} (Took {t_norm:.8f} seconds)")
    assert n_res == 5.0

    print("\n=================== DEMONSTRATING STATIC METHODS ===================")

    # Zeros Factory
    start = time.perf_counter()
    z_vec = Vec.zeros(5)
    t_z = time.perf_counter() - start
    print(f"Vec.zeros(5)   : {z_vec} (Took {t_z:.8f} seconds)")

    # Ones Factory
    start = time.perf_counter()
    o_vec = Vec.ones(5)
    t_o = time.perf_counter() - start
    print(f"Vec.ones(5)    : {o_vec} (Took {t_o:.8f} seconds)")

    # Uniform Random Factory
    start = time.perf_counter()
    r_vec = Vec.uniform(5)
    t_r = time.perf_counter() - start
    print(f"Vec.uniform(5) : {r_vec} (Took {t_r:.8f} seconds)")

    print("\nAll logical vector unit assertions passed successfully!")

if __name__ == "__main__":
    run_tests()
```

```python
%%writefile benchmark.py
import time
import numpy as np
import random
from vector_module import Vec

def run_benchmarks():
    print("\n=================== RUNNING ALL-OPERATION HIGH-DIMENSION BENCHMARKS ===================")
    # Scaled to exactly 6 specific dimensions from 2000 to 64000
    dimensions = [2000, 4000, 8000, 16000, 32000, 64000]
    iterations = 10

    for n in dimensions:
        print(f"\n" + "="*40 + f" DIMENSION: {n} " + "="*40)
        print(f"{'Operation Type / Method':<35} | {'Custom Pure Python (s)':<25} | {'NumPy Optimized (s)':<25}")
        print("-" * 91)

        # Pre-instantiate structures to keep benchmarking pure
        py_v1 = Vec.uniform(n)
        py_v2 = Vec.uniform(n)
        np_v1 = np.array(py_v1.elements)
        np_v2 = np.array(py_v2.elements)
        scalar = 2.5

        # ----------------------------------------------------
        # 1. Element-Wise Core Arithmetic
        # ----------------------------------------------------
        # Addition (__add__)
        start = time.perf_counter()
        for _ in range(iterations): _ = py_v1 + py_v2
        t_py_add = (time.perf_counter() - start) / iterations

        start = time.perf_counter()
        for _ in range(iterations): _ = np_v1 + np_v2
        t_np_add = (time.perf_counter() - start) / iterations
        print(f"{'Addition (__add__)':<35} | {t_py_add:<25.7f} | {t_np_add:<25.7f}")

        # Subtraction (__sub__)
        start = time.perf_counter()
        for _ in range(iterations): _ = py_v2 - py_v1
        t_py_sub = (time.perf_counter() - start) / iterations

        start = time.perf_counter()
        for _ in range(iterations): _ = np_v2 - np_v1
        t_np_sub = (time.perf_counter() - start) / iterations
        print(f"{'Subtraction (__sub__)':<35} | {t_py_sub:<25.7f} | {t_np_sub:<25.7f}")

        # Negation (__neg__)
        start = time.perf_counter()
        for _ in range(iterations): _ = -py_v1
        t_py_neg = (time.perf_counter() - start) / iterations

        start = time.perf_counter()
        for _ in range(iterations): _ = -np_v1
        t_np_neg = (time.perf_counter() - start) / iterations
        print(f"{'Negation (__neg__)':<35} | {t_py_neg:<25.7f} | {t_np_neg:<25.7f}")

        # ----------------------------------------------------
        # 2. Scalar Multiplication
        # ----------------------------------------------------
        # Left-Hand Scalar Multiplication (__rmul__)
        start = time.perf_counter()
        for _ in range(iterations): _ = scalar * py_v1
        t_py_rmul = (time.perf_counter() - start) / iterations

        start = time.perf_counter()
        for _ in range(iterations): _ = scalar * np_v1
        t_np_rmul = (time.perf_counter() - start) / iterations
        print(f"{'Scalar Multiplication (scalar * v)':<35} | {t_py_rmul:<25.7f} | {t_np_rmul:<25.7f}")

        # Right-Hand Addition Scenario (__radd__)
        start = time.perf_counter()
        for _ in range(iterations): _ = py_v2.__radd__(py_v1)
        t_py_radd = (time.perf_counter() - start) / iterations

        start = time.perf_counter()
        for _ in range(iterations): _ = np_v1 + np_v2
        t_np_radd = (time.perf_counter() - start) / iterations
        print(f"{'Right Addition (v2.__radd__(v1))':<35} | {t_py_radd:<25.7f} | {t_np_radd:<25.7f}")

        # ----------------------------------------------------
        # 3. In-Place Operations
        # ----------------------------------------------------
        # In-Place Addition (__iadd__)
        py_v_iadd = Vec.uniform(n)
        np_v_iadd = np.array(py_v_iadd.elements)

        start = time.perf_counter()
        for _ in range(iterations): py_v_iadd += py_v2
        t_py_iadd = (time.perf_counter() - start) / iterations

        start = time.perf_counter()
        for _ in range(iterations): np_v_iadd += np_v2
        t_np_iadd = (time.perf_counter() - start) / iterations
        print(f"{'In-Place Addition (v1 += v2)':<35} | {t_py_iadd:<25.7f} | {t_np_iadd:<25.7f}")

        # In-Place Multiplication (__imul__)
        py_v_imul = Vec.uniform(n)
        np_v_imul = np.array(py_v_imul.elements)

        start = time.perf_counter()
        for _ in range(iterations): py_v_imul *= scalar
        t_py_imul = (time.perf_counter() - start) / iterations

        start = time.perf_counter()
        for _ in range(iterations): np_v_imul *= scalar
        t_np_imul = (time.perf_counter() - start) / iterations
        print(f"{'In-Place Multiplication (v1 *= s)':<35} | {t_py_imul:<25.7f} | {t_np_imul:<25.7f}")

        # ----------------------------------------------------
        # 4. Reductions & Properties
        # ----------------------------------------------------
        # Euclidean Norm (norm)
        start = time.perf_counter()
        for _ in range(iterations): _ = py_v1.norm()
        t_py_norm = (time.perf_counter() - start) / iterations

        start = time.perf_counter()
        for _ in range(iterations): _ = np.linalg.norm(np_v1)
        t_np_norm = (time.perf_counter() - start) / iterations
        print(f"{'Euclidean Norm (norm())':<35} | {t_py_norm:<25.7f} | {t_np_norm:<25.7f}")

        # Dimensional Length (__len__)
        start = time.perf_counter()
        for _ in range(iterations * 100): _ = len(py_v1)
        t_py_len = (time.perf_counter() - start) / (iterations * 100)

        start = time.perf_counter()
        for _ in range(iterations * 100): _ = len(np_v1)
        t_np_len = (time.perf_counter() - start) / (iterations * 100)
        print(f"{'Vector Length (len(v))':<35} | {t_py_len:<25.7f} | {t_np_len:<25.7f}")

        # ----------------------------------------------------
        # 5. Factory Methods / Static Allocations
        # ----------------------------------------------------
        # Zeros Factory
        start = time.perf_counter()
        for _ in range(iterations): _ = Vec.zeros(n)
        t_py_zeros = (time.perf_counter() - start) / iterations

        start = time.perf_counter()
        for _ in range(iterations): _ = np.zeros(n)
        t_np_zeros = (time.perf_counter() - start) / iterations
        print(f"{'Zeros Factory (zeros(n))':<35} | {t_py_zeros:<25.7f} | {t_np_zeros:<25.7f}")

        # Ones Factory
        start = time.perf_counter()
        for _ in range(iterations): _ = Vec.ones(n)
        t_py_ones = (time.perf_counter() - start) / iterations

        start = time.perf_counter()
        for _ in range(iterations): _ = np.ones(n)
        t_np_ones = (time.perf_counter() - start) / iterations
        print(f"{'Ones Factory (ones(n))':<35} | {t_py_ones:<25.7f} | {t_np_ones:<25.7f}")

        # Uniform Distribution Factory
        start = time.perf_counter()
        for _ in range(iterations): _ = Vec.uniform(n)
        t_py_uni = (time.perf_counter() - start) / iterations

        start = time.perf_counter()
        for _ in range(iterations): _ = np.random.rand(n)
        t_np_uni = (time.perf_counter() - start) / iterations
        print(f"{'Uniform Random Factory (uniform(n))':<35} | {t_py_uni:<25.7f} | {t_np_uni:<25.7f}")

if __name__ == "__main__":
    run_benchmarks()
```
